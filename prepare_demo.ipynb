```python
import pandas as pd
import numpy as np

DATA_PATH = "creditcard.csv"
OUTPUT_PATH = "demo_transactions.csv"

df = pd.read_csv(DATA_PATH)

# Same feature engineering used during training
df["Hour"] = (df["Time"] / 3600) % 24

df["Hour_sin"] = np.sin(
    2 * np.pi * df["Hour"] / 24
)

df["Hour_cos"] = np.cos(
    2 * np.pi * df["Hour"] / 24
)

# Keep original Time because UI will display it
# Remove Hour because model doesn't use it
df = df.drop(columns=["Hour"])

# Select some genuine and fraud transactions
genuine = df[df["Class"] == 0].sample(
    10,
    random_state=42
)

fraud = df[df["Class"] == 1].sample(
    10,
    random_state=42
)

demo = pd.concat(
    [genuine, fraud]
).sample(
    frac=1,
    random_state=42
).reset_index(drop=True)

# Create demo ID
demo.insert(
    0,
    "Demo_ID",
    range(1, len(demo) + 1)
)

demo.to_csv(
    OUTPUT_PATH,
    index=False
)

print("Demo dataset created successfully.")
print("Shape:", demo.shape)
print("Saved as:", OUTPUT_PATH)
```
